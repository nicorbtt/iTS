```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
import datasets
import datetime
from ucimlrepo import fetch_ucirepo
```

For each dataset, after downolading (and in some cases processing) the data, we save them in two formats:
* a .csv file named `data.csv`, where each row stands for a different time series.
* two .json files, `train.json` and `test.json` respectively, that can be useful to prepare the data for [GluonTS](https://ts.gluon.ai/stable/api/gluonts/gluonts.torch.html) or [datasets](https://huggingface.co/docs/datasets/index)

```python
PATH = os.getcwd()
```

## Auto ##

taken from `https://github.com/canerturkmen/gluon-ts/tree/intermittent-datasets/datasets/intermittent_auto`

```python
PATH_AUTO = os.path.join(PATH, 'OnlineRetail')
if not os.path.exists(PATH_AUTO):
    os.makedirs(PATH_AUTO)
```

```python
#get .json from url

PATH_AUTO = os.path.join(PATH, 'Auto')
if not os.path.exists(PATH_AUTO):
    os.makedirs(PATH_AUTO)

test_url = 'https://raw.githubusercontent.com/canerturkmen/gluon-ts/intermittent-datasets/datasets/intermittent_auto/test/data.json'
train_url = 'https://raw.githubusercontent.com/canerturkmen/gluon-ts/intermittent-datasets/datasets/intermittent_auto/train/data.json'

train = pd.read_json(train_url)
test = pd.read_json(test_url)

print(len(train.target[0]), len(test.target[0]))
test
```

```python
#save the .csv train and test

test.to_json(os.path.join(PATH_AUTO, 'test.json'), orient='records')
train.to_json(os.path.join(PATH_AUTO, 'train.json'), orient='records')

pd.read_json(os.path.join(PATH_AUTO, 'train.json'), orient='records')
```

```python
#collect it into a .csv too

data = np.empty((len(test), 24))

for i, ts in enumerate(test.target):
    data[i,] = np.array(ts, dtype=np.int64)

pd.DataFrame(data).to_csv(os.path.join(PATH_AUTO, 'data.csv'), index=False)

data = pd.read_csv(os.path.join(PATH_AUTO, 'data.csv'))
data
```

```python
#save .json files

start = datetime.datetime(1996, 1, 1)

df_test = pd.DataFrame({
    'start' : [str(start) for i in range(len(data))],
    'target' : [list(data.iloc[i]) for i in range(len(data))],
    'feat_static_cat' : [[i] for i in range(len(data))],
    'feat_dynamic_cat' : [None for i in range(len(data))],
    'item_id' : ['T'+str(i) for i in range(len(data))]
})

df_train = pd.DataFrame({
    'start' : [str(start) for i in range(len(data))],
    'target' : [list(data.iloc[i])[:-12] for i in range(len(data))],
    'feat_static_cat' : [[i] for i in range(len(data))],
    'feat_dynamic_cat' : [None for i in range(len(data))],
    'item_id' : ['T'+str(i) for i in range(len(data))]
})

df_test.to_json(os.path.join(PATH_AUTO, 'test.json'), orient='records')
df_train.to_json(os.path.join(PATH_AUTO, 'train.json'), orient='records')

pd.read_json(os.path.join(PATH_AUTO, 'test.json'), orient='records')
```

```python
PATH_M5 = os.path.join(PATH, 'M5')
if not os.path.exists(PATH_M5):
    os.makedirs(PATH_M5)
```

```python
#import the data from train and test

sales_train_evaluation = pd.read_csv(os.path.join(PATH_M5, 'sales_train_evaluation.csv'), sep = ",").iloc[:,5:]
sales_test_evaluation = pd.read_csv(os.path.join(PATH_M5, 'sales_test_evaluation.csv'), sep = ",").iloc[:,5:]
sales_train_evaluation
```
